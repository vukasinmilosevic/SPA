1. Dodatni problemi iz sortiranja: Ukoliko nam j dat nesortirani niz, pronaci najmanju razliku izmedju bilo kog para u datom nizu.


Input: [1, 5, 3, 19, 18, 25]
Out: 1

```python
%%writefile exp1.cpp

#include <iostream>
#include <vector>
#include <climits>
#include <cmath>

using namespace std;

int minRaz(vector<int>& arr)
{
  // Vraca minimalnu razliku elemenata u nizu

  int n = arr.size();

  int diff = INT_MAX;

  // Petlje koje nam racunaju razlike i uporedjuju sa diff

  for (int i = 0; i < n-1; i++)
  {
    for (int j = i+1; j < n; j++)
    {
      if (abs(arr[i] - arr[j]) < diff)
        diff = abs(arr[i] - arr[j]) ;
    }

  }

 return diff;
}
int main()
{

  vector<int> arr = {1, 5, 3, 19, 18, 25};

  cout << minRaz(arr);
  return 0;

}
```

2. Imamo set novcica sa vrednostima [1, 2, 5, 10] i potrebno je da date minimalni broj novcica nekome za kusur od 39.

Greedy algortiam prisputa problemu:
- Pocinjemo sa najvecim novcicem koji je <= of vrednosti koju treba da vratimo - u nasem primeru je to 10.
- Oduzimamo najveci novcic, njegovu vrednost, od sume: 39-10 = 29 i dodajemo jedan novcic od 10 u nase resenje.

Ponavljamo proces dok suma ne postane 0.

```python
%%writefile exp2.cpp

#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

vector<int> minNovcic(vector<int>& novcic, int suma)
{

  int n = novcic.size();
  sort (novcic.begin(), novcic.end());
  vector<int> res(novcic.size(),0) ;

  //Pocinjemo od novcica sa najvecom vrednoscu

  for (int i = n-1; i>=0; i--)
  {
     if (suma >= novcic[i]){
        //Trazimo maksimalni broj novcica te vrednosti koji mozemo koristiti

        int brojac = (suma / novcic[i]);
        res[i] = brojac;
        suma -= (brojac * novcic[i]);

     }

    //Zaustavi racunicu ukoliko je suma = 0

    if (suma == 0)
       break;

  }

  return res;

}
int main()
{

  vector<int> novcic = {5, 2, 1, 10};
  int suma = 39;

  vector<int> rez = minNovcic(novcic, suma);
  sort(novcic.begin(),novcic.end());
  cout<<"Koristili smo:"<<endl;
  for (int i = 0; i < rez.size(); i++)
      cout<<rez[i]<<" x "<< novcic[i]<< " din"<<endl;

  return 0;

}
```

3. Pravljenje plana rada u firmi:

Ukoliko imamo dva niza, rok[] i profit[], pri cemu rok[i] predstavlja poslednju vremensku jedinicu do koje i-ti zadatak mora da bude kompletiran, a profit[i] predstavlja profit koji firma zaradi ukoliko je i-ti zadatak kompletiran. Izvrsavanje svakog zadatka traje 1 vremensku jedinicu i samo jedan posao moze biti obavljan u jednom trenutku. Pronaci broj zavrsenih zadataka i profit.

```python
%%writefile exp3.cpp

#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

vector<int> pravljenjeRasporeda(vector<int>& rok, vector<int>& profit)
{
  int n = rok.size();
  int brojac = 0;
  int totProfit = 0;


  //Uparimo profit i rok za sve poslove zajedno

  vector<pair<int, int>> jobs;

  for (int i = 0; i < n; i++)
  {
    jobs.push_back({profit[i], rok[i]});
  }

  //Sortiramo zadatke po profitu u opadajucem redu

  sort(jobs.begin(), jobs.end(), greater<pair<int, int>>());

  vector<int> slot(n, 0);

  for (int i = 0; i < n; i++){

    int start = min(n, jobs[i].second)-1;
    for (int j = start; j>=0; j--){

        //ukolikoje slot prazan
        if (slot[j] == 0) {
          slot[j] = 1;
          brojac++;
          totProfit+= jobs[i].first;
          break;
        }

    }

  }
  return {brojac, totProfit};
}
int main()
{
  vector<int> rok = {2,1,2,1,1};
  vector<int> profit = {100, 19, 27, 25, 15};
  vector<int> odgovor = pravljenjeRasporeda(rok, profit);
  cout<<odgovor[0]<<" "<<odgovor[1];

  return 0;

}
```
