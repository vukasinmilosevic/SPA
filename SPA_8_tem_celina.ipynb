8. Uvod u Tree Data strukturu

```python
%%writefile tree_def.cpp

#include <iostream>
#include <vector>
using namespace std;

//Struktora node za Tree

class Node{
public:
    int data;
    vector<Node*> children;

    Node(int x){
        data = x;
    }

};

// Fukncija da dodamo child u node-u

void addChild(Node* parent, Node* child){
    parent->children.push_back(child);
}

// Funkcja koja print-uje roditelje za svaku node

void printParents(Node* node, Node* parent){
    if (parent == nullptr)
      cout << node->data << "-> NULL" << endl;
    else
      cout << node->data << "->" << parent->data << endl;

    for (auto child : node->children)
      printParents(child, node);
}

// Funkcija koja print-uje decu svake node

void printChildren(Node* node){
  cout << node->data << " -> ";

  for (auto child: node->children)
    printChildren(child);
}

// Funkcija koja print-uje lead node

void printLeafNodes(Node* node){
  if (node->children.empty()){
    cout << node->data << "";
    return;
  }
  for (auto child: node->children)
    printLeafNodes(child);
}

// Fukcija koja print-uje degrees svake node

void printDegrees(Node* node, Node* parent){
  int degree = node->children.size();
  if (parent != nullptr)
    degree++;
  cout<< node->data << " -> " << degree << endl;

  for (auto child: node->children)
    printDegrees(child, node);
}

int main(){

  // Kreiramo nase node

  Node* root = new Node(1);
  Node* n2 = new Node(2);
  Node* n3 = new Node(3);
  Node* n4 = new Node(4);
  Node* n5 = new Node(5);

  // Konstruisemo nase Tree

  addChild(root,n2);
  addChild(root, n3);
  addChild(n2, n4);
  addChild(n2, n5);

  cout << "Roditelji svake node: " <<endl;
  printParents(root, nullptr);

  cout<< "Deca svake node: "<< endl;
  printChildren(root);

  cout<< "Leaf nodes: ";
  printLeafNodes(root);
  cout<<endl;

  cout<< "Degrees of node: "<<endl;
  printDegrees(root, nullptr);

  return 0;

}
```

```python
!g++ tree_def.cpp -o tree_1
!./tree_1
```

```python
 #8.1. Inorder traversal of binary tree

%%writefile inOrder_exp.cpp

#include <iostream>
#include <vector>
using namespace std;

//Struktora node za Tree

class Node{
public:
    int data;
    Node* left;
    Node* right;

    Node(int x){
        data = x;
        left = right = NULL;
    }

};

void inOrder(Node* node, vector<int>& res){
  if (node == nullptr)
    return;

  // Prolazak kroz levi subtree

  inOrder(node->left, res);

  res.push_back(node->data);

  inOrder(node->right, res);
}



int main(){

  // Kreiramo nase node, koristimo binary tree
  //          1
  //         / \
  //        2   3
  //       / \   \
  //      4   5   6
  //         / \
  //        7   8


  Node* root = new Node(1);
  root->left = new Node(2);
  root->right = new Node(3);
  root->left->left = new Node(4);
  root->left->right = new Node(5);
  root->left->right->left = new Node(7);
  root->left->right->right = new Node(8);
  root->right->right = new Node(6);

  vector<int> res;
  inOrder(root, res);

  for (int node: res)
    cout << node << "";


  return 0;

}
```

```python
 #8.3. Postorder traversal of binary tree

%%writefile postOrder_exp.cpp

#include <iostream>
#include <vector>
using namespace std;

//Struktora node za Tree

class Node{
public:
    int data;
    Node* left;
    Node* right;

    Node(int x){
        data = x;
        left = right = NULL;
    }

};

void inOrder(Node* node, vector<int>& res){
  if (node == nullptr)
    return;

  // Prolazak kroz levi subtree

  inOrder(node->left, res);

  inOrder(node->right, res);

  res.push_back(node->data);

}



int main(){

  // Kreiramo nase node, koristimo binary tree
  //          1
  //         / \
  //        2   3
  //       / \   \
  //      4   5   6
  //         / \
  //        7   8


  Node* root = new Node(1);
  root->left = new Node(2);
  root->right = new Node(3);
  root->left->left = new Node(4);
  root->left->right = new Node(5);
  root->left->right->left = new Node(7);
  root->left->right->right = new Node(8);
  root->right->right = new Node(6);

  vector<int> res;
  inOrder(root, res);

  for (int node: res)
    cout << node << "";


  return 0;

}
```

```python
 #8.4. Level order traversal of binary tree

%%writefile lvlOrder_exp.cpp

#include <iostream>
#include <vector>
using namespace std;

//Struktora node za Tree

class Node{
public:
    int data;
    Node *left;
    Node *right;

    Node(int x){
        data = x;
        left = right = nullptr;
    }

};

void lvlOrder(Node* root, int level, vector<vector<int>>& res){
  if (root == nullptr) return;

  // Dodaj novi nivo u rezulat, ako je to potrebno
  if (res.size() <= level)
      res.push_back({});

  // Dodaj podatke iz trenutne node na odgovarajuci nivo

  res[level].push_back(root->data);

  //Rekurzvno za svu levu i desnu decu

  lvlOrder(root->left, level+1, res);
  lvlOrder(root->right, level+1, res);

}

// Funkcija koja vrsi lvl order travesal

vector<vector<int>> lvlOrderfinal(Node *root){
  // Cuvamo rezultate nivo po nivo
  vector<vector<int>> res;

  lvlOrder(root, 0, res);
  return res;
}

int main(){

  // Kreiramo nase node, koristimo binary tree
  //          1
  //         / \
  //        2   3
  //       / \   \
  //      4   5   6
  //         / \
  //        7   8


  Node* root = new Node(1);
  root->left = new Node(2);
  root->right = new Node(3);
  root->left->left = new Node(4);
  root->left->right = new Node(5);
  root->left->right->left = new Node(7);
  root->left->right->right = new Node(8);
  root->right->right = new Node(6);

  vector<vector<int>> res = lvlOrderfinal(root);

  for (vector<int> level: res)
  {
    cout<<endl;
    for (int val: level)
      cout<< val<< " ";
}
    cout<<endl;


  return 0;

}
```
