4. Donja granica

Ukoliko nam je dat sortiran niz arr[] i broj target, pronaci donju granicu target-a u ovom datom nizu. Donja granica je broj koji je definisan kao najmanji indeks u tom sortiranom nizu gde je element veci ili jednak target-u. Napomena: Ukoliko su svi elementi niza manji od target-a, donja granica je jednaka duzini niza.

Primer:

Input: arr[] = [2,3,7,10,11,11,25], target = 9
Output: 3

```python
%%writefile problem5.cpp

//CPP test program

#include <iostream>
#include <vector>
using namespace std;

int pretraga(vector<int>& arr, int x){
  // Vrsimo iteraciju po nizu da bismo pronasli donju granicu x

  int n = arr.size();

  for (int i=0; i<n; i++)
    if (arr[i] >= x)
      return i;

  return n;

}

int main(){

vector<int> arr = {2,3,7,10,11,11,25};
int x = 9;
int rezultat = pretraga(arr, x);


cout <<rezultat;

return 0;


}
```
