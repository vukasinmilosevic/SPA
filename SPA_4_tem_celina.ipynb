```python
%%writefile bubble_sort_exp1.cpp

#include <iostream>
#include <vector>

using namespace std;

// Optimizovana verzija bubble sort-a

void bubbleSort(vector<int>& arr){

  int n = arr.size(); // n = 4
  bool swapped;

  for (int i=0; i < n-1; i++){
    /*

     i = 0
     swapped = false
     j = n - i - 1 = 4-1 = 3

     j = [0,1,2,3]


    */
    swapped = false;
    for (int j = 0; j < n-i-1; j++){

      if (arr[j]>arr[j+1]){
          swap(arr[j], arr[j+1]);
          swapped = true;
      }

    }

    for (int num : arr)
      cout<< num <<"";
    cout<<endl;
    //Ukoliko niti jedna dva elementa nisu zamenjena break
    if (!swapped)
      break;

  }

}


int main(){
   vector<int> arr = {5,6,1,3};
   bubbleSort(arr);
   for (int num : arr)
      cout<< num <<"";

}
```

```python
%%writefile insertion_sort_exp2.cpp

#include <iostream>
#include <vector>

using namespace std;

// Insertion sort

void insertionSort(vector<int>& arr, int n) {

  for (int i = 1; i<n; ++i){
    int key = arr[i];
    int j = i-1;
    // pomeramo elemente vece od key za jednu poziciju napre
    while (j>=0 && arr[j] > key){
      arr[j+1] = arr[j];
      j--;
    }
    arr[j+1] = key;
    for (int num : arr)
      cout<< num <<" ";
      cout<<endl;
  }
}
int main(){
   vector<int> arr = {12,11,13,5,6};
   insertionSort(arr, arr.size()); // Changed 4 to arr.size()
   for (int num : arr)
      cout<< num <<" ";
}
```

```python
%%writefile selection_sort_exp3.cpp

#include <iostream>
#include <vector>

using namespace std;

void selectionSort(vector<int> &arr){
  int n = arr.size();
  for (int i = 0; i<n-1; i++){
    //pretpostavka je da trenutni iterator/pozicija usmerava ka minimalnom elementu
    int min_idx = i;
    //iteracija kroz nesortirani deo niza zarad pronalaska stvarnog minimuma
    for (int j = i+1; j< n; ++j){
      if (arr[j]<arr[min_idx]){
        //update-ovanje min_idx ukoliko smo nasli element manji od onog na poziciji min_idx
        min_idx = j;
      }
    }
    //pomeramo minimalni element na njegovu trenutnu poziciju
    swap(arr[i], arr[min_idx]);

    for (int num : arr)
      cout<< num <<" ";
    cout<<endl;
  }
}
int main(){
   vector<int> arr = {64,25,12,22,11};
   selectionSort(arr);
   for (int num : arr)
      cout<< num <<" ";
}


```

1. Sortirati niz string-ova shodno njihovoj duzini

Primer:
* Input: ["student", "ja", "sam"]
* Ouput: ["ja", "sam", "student"]



```python
%%writefile bubble_sort_exp2.cpp

#include <iostream>
#include <vector>

using namespace std;

// Optimizovana verzija bubble sort-a

void bubbleSort(vector<string>& arr){

  int n = arr.size(); // n = 4
  bool swapped;

  for (int i=0; i < n-1; i++){
    swapped = false;
    for (int j = 0; j < n-i-1; j++){

      if (arr[j].size()>arr[j+1].size()){
          swap(arr[j], arr[j+1]);
          swapped = true;
      }

    }

    //Ukoliko niti jedna dva elementa nisu zamenjena break
    if (!swapped)
      break;

  }

}

int main(){
   vector<string> arr = {"student", "ja", "sam"};
   bubbleSort(arr);
   for (string num : arr)
      cout<< num <<" ";

}
```

```python
%%writefile wave_exp2.cpp

#include <iostream>
#include <vector>

using namespace std;



void sortInWave(vector<int>& arr){
    int n = arr.size();
    for(int i=0; i< n-1; i+=2)
    {
      swap(arr[i], arr[i+1]);
    }
}

int main(){
   vector<int> arr = {1,2,3,4,5};
   sortInWave(arr);
   for (int num : arr)
      cout<< num <<" ";
}
```

```python
%%writefile exp3.cpp
// Za niz intervala, vratiti spojene intervale, koji su ortogonalni
//Input: [ [1,3], [2,4], [6,8], [9,10] ]
//Output: [ [1,4], [6,8], [9,10] ]

#include <iostream>
#include <vector>
#include <bits/stdc++.h>

using namespace std;

vector<vector<int>> rangeMerge(vector<vector<int>>& arr){
    int n = arr.size();
    sort(arr.begin(), arr.end());
    vector<vector<int>> rez;
    for(int i=0; i < n; i++)
    {
      int start = arr[i][0];
      int end = arr[i][1];
      //preskociti vec spojene intervale

    if (!rez.empty() && rez.back()[1]>=end)
        continue;
    // Ponaci kraj spojenog raspona
    for (int j=i+1; j<n; j++){
      if (arr[j][0]<=end)
        end = max(end, arr[j][1]);
    }
    rez.push_back({start, end});
    }
  return rez;
}

int main(){
   vector<vector<int>> arr = { {1,3}, {2,4}, {6,8}, {9,10} };
   vector<vector<int>> rez = rangeMerge(arr);
   for (auto num : rez)
      cout<< num[0] <<" "<<num[1]<<endl;
}
```
