1. Simple implementation of queue using arrays

```python
%%writefile queue_implementation_example.cpp

#include <iostream>
using namespace std;

class myQueue {

  // Niz u kojem cemo cuvati elemente
  int *arr;

  // Maksimalni broj elemenata u queue
  int capacity;

  // Trenutni broj elemenata u nizu.
  int size;

  public:

    myQueue(int c) {
      capacity = c;
      arr = new int[capacity];
      size = 0;
    }

    bool isEmpty()
    {
      return size == 0;
    }

    bool isFull()
    {
      return size == capacity;
    }

    // Dodavanje elementa x na kraj queue

    void enqueue(int x)
    {
      if (isFull())
      {
        cout <<"Queue je pun!\n";
        return;
      }
      arr[size] = x;
      size++;
    }

    //Brisanje front elementa unutar queue

    void dequeue()
    {

      if (isEmpty())
      {
        cout << "Queue je prazan!\n";
        return;
      }
      for (int i = 1; i<size ; i++)
      {
        arr[i-1] = arr[i];
      }

      size--;
    }

    //Vracanje front elementa queue-a

    int getFront()
    {

      if (isEmpty())
      {
        cout << "Queue je prazan!\n";
        return -1;
      }
      return arr[0];

    }
   //Vracaje poslednjeg elementa queue-a

   int getRear()
   {
      if (isEmpty())
      {
        cout << "Queue je prazan!\n";
        return -1;
      }
      return arr[size-1];
   }

};


int main()
{
  myQueue q(3);

  q.enqueue(10);
  q.enqueue(20);
  q.enqueue(30);

  cout<< "Front: "<<q.getFront()<<endl;
  cout<< "Rear: "<<q.getRear()<<endl;

  q.enqueue(40);
  q.dequeue();

  cout<< "Front: "<<q.getFront()<<endl;
  cout<< "Rear: "<<q.getRear()<<endl;

  return 0;

  }
```



2. Implementation of circular queue using Array


```python
%%writefile queue_implementation_example2.cpp

#include <iostream>
using namespace std;

class myQueue {

private:
  // fixed size array
  int* arr;

  // index front elementa;

  int front;

  //trenutni broj elemenata

  int size;

  // maximum capacity

  int capacity;

public:

  myQueue(int cap){

      capacity = cap;
      arr = new int[capacity];
      front = 0;
      size = 0;
  }


  //Insert elemnta x na kraj

  void enqueue(int x){
      if (size == capacity)
      {
        cout<<"Queue je pun!"<<endl;
        return;
      }

      int rear = (front + size) % capacity;
      arr[rear] = x;
      size++;

  }

  //Brisanje elementa sa kraja

  int dequeue()
  {
    if (size == 0)
      {
        cout<<"Queue je prazan!"<<endl;
        return -1;
      }

      int res = arr[front];
      front = (front + 1) % capacity;
      size--;
      return res;
  }

// Vracanje front elementa

int getFront()
{
  if (size == 0) return -1;
  return arr[front];
}
// Vracanje rear elementa
int getRear()
{
  if (size == 0) return -1;
  int rear = (front + size - 1) % capacity;
  return arr[rear];

}

void print_array(){

  cout<< "Printing arr: ";
  for (int i = 0; i < capacity; i++)
    cout<< arr[i] << ", ";

  cout<<"Kraj ;"<<endl;
}

};



int main(){

    myQueue q(5);

    q.enqueue(10);
    q.enqueue(20);
    q.enqueue(30);

    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;
    q.print_array();
    q.dequeue();
    q.print_array();

    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;

    q.enqueue(40);
    q.print_array();

    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;

    q.enqueue(50);
    q.print_array();

    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;

    q.enqueue(60);
    q.print_array();

    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;

    q.enqueue(70);

    q.dequeue();
    q.print_array();
    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;

    q.enqueue(70);
    q.print_array();

    cout<< "Front: "<<q.getFront()<<endl;
    cout<< "Rear: "<<q.getRear()<<endl;

    return 0;
}
```
