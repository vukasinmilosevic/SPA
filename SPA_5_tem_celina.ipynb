1. Ukoliko zelimo da postavimo odredjeni bit na n-toj poziciji unutar broja "num", to moze biti uradjeno koriscenjem "OR" operatora (|).
* Zelimo da pomirmo sa left shit 1 na n-tu poziciju (1 << n)
* Koristimo "OR" operator da podesimo bit na toj poziciji.

```python
%%writefile p1.cpp

#include <bits/stdc++.h>
#include <iostream>
using namespace std;

// num - broj, pos - pozicija na kojoj zelimo da dodamo bit

void Set(int& num, int pos)
{
  // Prvi korak nam je da shift-ujemo 1,
  //drugi korak je bitwise OR

  num |= (1 << pos);

}

int main()
{
  int num = 4, pos = 1;
  Set(num, pos);
  cout<< num << endl;
  return 0;

}
```

```python
!g++ p1.cpp -o p1
!./p1
```
