##Algoritmi

Algoritam je skup konačnih, jasno definisanih koraka ili instrukcija namenjenih rešavanju problema ili izvršavanju proračuna. Može se definisati i kao postupak za rešavanje matematičkog ili računarskog problema u konačnom broju koraka, često uz ponavljanje ili rekurzivne operacije.

###Potreba za algoritmima:

Efikasno i delotvorno rešavanje složenih problema.
Automatizacija procesa, čineći ih pouzdanim, bržim i jednostavnijim.
Omogućavaju računarima da izvršavaju zadatke koji su teški ili nemogući za ljude.
Široko se koriste u matematici, računarstvu, eksperimentalnoj fizici, inženjerstvu, finansijama i uopšteno u analizi podataka.

###Upotreba algoritama
Algoritmi su ključni za efikasno rešavanje problema u različitim oblastima:

**Računarstvo**: Osnova programiranja, od jednostavnog sortiranja i pretrage do veštačke inteligencije i mašinskog učenja.

**Matematika**: Rešavanje problema poput linearnih jednačina, najkraćih puteva i optimizacije.

**Operaciona istraživanja:** Optimizacija logistike, transporta i raspodele resursa.

**Veštačka inteligencija:** Pokreću inteligentne sisteme za zadatke kao što su prepoznavanje slika, obrada prirodnog jezika i donošenje odluka.

**Nauka o podacima:** Analiza i izvlačenje uvida iz velikih skupova podataka u marketingu, zdravstvu, finansijama itd.

**Algoritmi mogu biti jednostavni ili složeni, u zavisnosti od zadatka. Možete ih zamisliti kao recept: korak po korak dovodi do željenog rezultata.**

###Osobine algoritma
Algoritam je precizan skup instrukcija za rešavanje problema. Da bi skup instrukcija bio algoritam, mora imati sledeće osobine:

**Jasnoća i nedvosmislenost:** Svaki korak mora biti precizan i imati samo jedno značenje.

**Dobro definisani ulazi:** Algoritam može imati nula ili više ulaza, koji moraju biti jasno određeni.

**Dobro definisani izlazi:** Mora proizvesti bar jedan jasno definisan rezultat.

**Konačnost:** Algoritam se mora završiti nakon konačnog broja koraka.
Beskonačne petlje ili rekurzija bez uslova prekida krše ovu osobinu.

**Efikasnost (izvodljivost):**Koraci treba da budu jednostavni, izvodljivi i mogući uz dostupne resurse.

**Determinističnost:** Za isti ulaz algoritam uvek daje isti izlaz.

**Nezavisnost od jezika:** Algoritmi su opšte instrukcije i mogu se implementirati u bilo kom programskom jeziku uz isti rezultat.

###Kako izraziti algoritam?

**Prirodni jezik:** Pisanje običnim jezikom. Lako za opis, ali može biti nejasno kod složenih problema.

**Dijagram toka (flowchart):** Grafički prikaz koraka. Lakši za vizuelno razumevanje.

**Pseudokod:** Tekstualni prikaz sličan kodu, bez striktne sintakse programskog jezika.


**Najbolji način izražavanja algoritma:** Treba da bude razumljiv čak i početnicima sa osnovnim znanjem.

###Koraci za dizajniranje algoritma

Pri dizajniranju algoritma treba uzeti u obzir:

* Definiciju problema: Jasno odrediti šta se rešava.
* Ograničenja: Uočiti pravila i ograničenja.
* Ulaze: Odrediti koje podatke algoritam prima.
* Izlaze: Definisati očekivane rezultate.
* Izvodljivost rešenja: Proveriti da li rešenje funkcioniše u datim uslovima.

1. Dat nam je broj n. Potrebno je proveriti da li je on paran ili neparan. Vratiti ```true``` ukoliko je paran, a ```false``` ako je neparan.
```
Input n=15
Output false

Hint - 15%2 = 1, znaci 15 je neparan
```

````python
%%writefile main.cpp
#include <iostream>
using namespace std;

bool Paran(int n)
{

  int ostatak = n % 2;
  if (ostatak == 0)
  {
    return true;
  }
  else{
    return false;
  }

}

int main()
{

  int n = 15;
  if (Paran(n))
    cout<< "true";
  else
    cout<< "false";

  return 0;

}
````

3. Prikazati sumu prvih n prirodnih brojeva (za dato n).
```
Input n = 3

Output 6
Hint: 1+2+3 = 6
```

````python
%%writefile main3.cpp
#include <iostream>
using namespace std;

int PronaciSumu(int n)
{
  int suma = 0;
  for (int i = 1; i<= n; i++)
  {
    suma = suma + i ;
  }
  return suma;
}

int main()
{
  int n = 3;
  cout<< PronaciSumu(n);
  return 0;
}
````

5. Zameniti (swap) dva broja a i b.
```
Input: a = 2, b = 3
Output: a = 3, b = 2
```

````python
%%writefile main5.cpp
#include <iostream>
using namespace std;


int main()
{
  int a = 2;
  int b = 3;

  int tmp = b;
  b = a;
  a = tmp;

  cout<< "a = "<< a << endl;
  cout << "b = "<< b << endl;

  return 0;
}
````

6. Ukoliko su nam data dva int n i m (m ne moze biti 0), pronaci broj koji je najblizi n, a da je pritom deljiv sa m. Ukoliko postoji vise od jednog broja, tada treba vratiti broj koji ima maksimalnu apsolutnu vrednost.

```
Input: n = 13, m = 4
Output: 12
```

````python
%%writefile main6.cpp
#include <iostream>
#include <climits>
using namespace std;

int closestNumber(int n, int m){

  int najblizi = 0;
  int minRazlika = INT_MAX;

  //Proverimo brojeve u okolini n
  for (int i = n - abs(m); i <=n+abs(m); ++i)
  {
    if (i % m == 0){
      int razlika = abs(n-i);

      if (razlika < minRazlika || (razlika == minRazlika && abs(i)>abs(najblizi) ))
      {
        najblizi = i;
        minRazlika = razlika;
      }
    }
  }
  return najblizi;
}

int main(){

  int n = 13, m = 4;
  cout<< closestNumber(n,m)<<endl;

  return 0;

}
````
